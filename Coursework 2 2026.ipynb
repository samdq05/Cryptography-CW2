Complete the definition of the Python function `divides_frob` below. It takes a non-constant polynomial $f \in \mathbb F_2[X]$ and returns `True` if $f$ divides the polynomial $X^{2^n} - X$, where $n = \deg(f)$. Otherwise it returns `False`.

There is a naïve but highly inefficient way to write the function. Ensure your implementation is efficient.

### Plan:

$f \vert (X^{2^n} - X) \implies X^{2^n} - X \equiv 0 (\text{mod} f) \implies X^{2^n} \equiv X \, (\text{mod} f )$ as it's in $\mathbb{F}_2$, $-X \equiv X (\text{mod} f)$.

So, rather than computing $X^{2^n} - X$ and checking divisibility (bad as $2^n$ is generally large), I can use repeated doubling on $X^{2^n}$ and compare to $X (\text{mod} f)$ as shown above.

````python
def divides_frob(f):
    """Determines if the non-constant polynomial f in GF(2)[X] divides X^{2^n} - X, where n = deg(f).
    Equally, if X^{2^n} = X (mod f)"""
    n = f.degree()
    F = GF2QuotientField(f)
    g = F(0b10)
    for i in range(n):
        g = g**2  # as it's in F it automatically reduces modulo f
    return g == F(0b10)
````

Complete the definition of the function `distinct_prime_factors` below so that for each integer $n \ge 2$, `distinct_prime_factors(n)` is a list of the distict prime factors of $n$, listed in order of increasing size. Thus, `distinct_prime_factors(12)` should be `[2, 3]`.

````python
def distinct_prime_factors(n):
    """Gives a list of distinct primes whose product is n"""
    d = 2
    listo = []
    while d*d <= n:
        if n % d == 0:    
            listo.append(d)   # by appending beforehand, I ensure distinct prime factors
        while n % d == 0:
            n = n //d
        d = d + 1
    if n == 1:
        return listo
    else:
        listo.append(n)
        return listo
````

### Plan:

3. Compute $X^{2^n_i} (\text{mod} f)$ as the gcd calculation yields the same result, but now we are working with smaller polynomials
4. Add $X$ (equivalent to subtracting in $\mathbb{F}_2$)
5. Compute $\text{gcd}(f, X^{2^n_i} - X)$ using the gcd function
6. Reutrn `False` if any gcd isn't 1, otherwise return `True` 

````python
def all_gcd_frob(f):
    """Given a polynomial f in GF2[X] of degree n >= 2. Let p_1, ... p_k be the distinct prime factors of n.
    Let n_i = n / p_i (for i = 1, ..., k). This function returns True if gcd(f, X^(2^n_i) - X) = 1 for 
    i = 1, ..., k. Otherwise, it returns False."""
    # 1.
    n = f.degree()
    ps = distinct_prime_factors(n)
    for p in ps:
        # 2.
        ni = n // p
        # 3.
        F = GF2QuotientField(f) 
        g = F(0b10) 
        for i in range(ni):
            g = g**2
        # 4.
        g = g + F(0b10) 
        # 5. & 6. 
        if f.gcd(g.poly) != GF2Polynomial(0b1):
            return False
    return True
````

### Plan:

The primitive root test says: Let $p$ be an odd prime, and $a \in \mathbb{Z}, p \nmid a$. Then $a$ is a primitive root

of $p − 1, a^{\frac{p−1}{q}} \not\equiv 1 \text{mod} p$. 

If $a$ is a primitive root modulo $p$, then it is a generator of the unit group $(\mathbb{Z}/p\mathbb{Z})^\times$.

````python
def is_generator(f):
    """Determines whether f a generator of the unit group of the finite field to which it belongs"""
    if not isinstance(f, GF2FieldElement): # ensure f is an element of a finite field
        raise TypeError("is_generator expects a field element")
    F = f.field # the finite field to which f belongs
    one = F(1) # the one polynomial in the finite field.
    n = field_order(F) - 1
    factor_list = distinct_prime_factors(n)
    for q in factor_list:
        if f**(n // q) == one:
            return False
    return True
````

## Question 3(d) [6 marks]

Complete the definition of the function `bsgs` below. It should take non-zero arguments `g` and `h` that belong to a finite field (technically, that are instances of `GF2FieldElement`), and belong to the same field. You may assume `g` is a generator of the unit group of the quotient field to which both $g$ and $h$ belong. The function returns a non-negative integer `z` such that $g^z = h$.

Your function should implement the baby-step giant-step method. I suggest creating a dictionary for the baby step list where the key is the group element and the value is the exponent. You can save time and space by not storing the giant list. Instead, iterate through the elements of the list, checking if each element is in the baby list.

### Plan:

1. Set $n = \lceil \sqrt{N} \rceil$, smallest integer that is at least $\sqrt{N}$, where $N = \text{ord}(g) = 2^{\text{degree}} -1$, as $g$ is a generator of the unit group 
2. Construct a list $e, g, g^2, ..., g^{n-1}$.
3. Construct a list $h, hg^{-n}, hg^{-2n},..., hg^{-(n-1)n}$. But as negative powers won't work, I can use Fermat's little theorem: since $g^N = 1$ in the group $\implies g^{-1} = g^{N-1} \implies g^{-n} = g^{N - n}$, which is non-negative as $n \leq N$.
4. Search for a match between the lists. That is, find $0 \leq i < n$ and $0 \leq j < n$ such that $g^i = hg^{-jn}$.
5. Then $h = g^{i+jn}$ so $x \leftarrow i + jn$ is a solution.

````python
def bsgs(g, h):
    """Given that g and h are both non-zero elements of the same finite field, this function uses
    baby-step giant-step to return z such that g^z = h, if such a z exists"""
    if not isinstance(g,  GF2FieldElement) or not isinstance (h, GF2FieldElement):
        raise TypeError(f"{g} and {h} must be field elements")
    F = g.field
    if F != h.field:
        raise ValueError(f"{g} and {h} must belong to the same field")
    if g == 0 or h == 0:
        raise ValueError(f"both arguments {g} and {h} must be non-zero")

    # 1
    deg = g.field.modulus.degree()
    N = 2**deg - 1
    n = ceil(sqrt(N))
    
    # 2
    baby_dict = {}
    for i in range(n):
        baby_dict[g**i] = i  # baby step

    # 3
    current_giant = h
    giant_step = g**(N-n) # g^(-n) = g^(N-n)
    for j in range(n):    
        # 4 check for match   
        if current_giant in baby_dict:
            return baby_dict[current_giant] + j*n # 5
        current_giant = current_giant * giant_step  # giant step
     
    return None
````

## Question 4(a) [3 marks]

In Question 3, you wrote a baby-step giant-step algorithm to solve the DLP $g^x =h$ where you assumed that $g$ is a generator of the unit group of the finite field to which both $g$ and $h$ belong. The algorithm is not so efficient if $g$ and $h$ belong to some proper subgroup of this group.

Below, complete the definition of the function `bsgs_sub` where `bsgs_sub(g, h, N)` solves the same discrete logarithm problem but where `g` is assumed to be a generator of an order `N` subgroup of the unit group of the finite field and where `h` is assumed to be an element of this subgroup. This should involve only a small modification of your previous code.

Add an appropriate docstring (the string between triple quotation marks) to describe the function.

````python
def bsgs_sub(g, h, N):
    """Given that g and h are both non-zero elements of the same finite field and N is assumed to be 
    the order of the subgroup generated by g, this function uses baby-step giant-step to return z such that 
    g^z = h, if such a z exists. Returns None if no such z exists."""
    if not isinstance(g,  GF2FieldElement) or not isinstance (h, GF2FieldElement):
        raise TypeError(f"{g} and {h} must be field elements")
    F = g.field
    if F != h.field:
        raise ValueError(f"{g} and {h} must belong to the same field")
    if g == 0 or h == 0:
        raise ValueError(f"both arguments {g} and {h} must be non-zero")
    if not isinstance(N, int) or N < 1:
        raise ValueError("N must be a positive integer")

    # 1
    n = ceil(sqrt(N))
    
    # 2
    baby_dict = {}
    for i in range(n):
        baby_dict[g**i] = i  # baby step

    # 3
    current_giant = h
    giant_step = g**(N-n) # g^(-n) = g^(N-n)
    for j in range(n):    
        # 4 check for match   
        if current_giant in baby_dict:
            return baby_dict[current_giant] + j*n # 5
        current_giant = current_giant * giant_step  # giant step
     
    return None
````

Let $n \ge 2$ be an integer. Complete the definition of the function `prime_power_factors` below such that `prime_power_factors(n)` is a list of all the prime power factors of $n$, in increasing order.

For example, `prime_power_factors(200)` should be `[8, 25]`. You may factorise via trial division.

````python
def prime_power_factors(n):
    """Gives a list of the prime power factors of n"""
    d = 2
    dicto = {}
    while d*d <= n:
        while n % d == 0:
            if d not in dicto:
                dicto[d] = 0
            dicto[d] += 1
            n = n //d
        d = d + 1
    if n == 1:
        return [p**k for p, k in dicto.items()]
    else:
        dicto[n] = 1
        return [p**k for p, k in dicto.items()]
````

## Question 4(c)(i) [6 marks]

Complete the definition of the the function `pohlig_hellman` below. Here, `f` is a discrete log solving algorithm such that `f(a, b, N)` returns `x` such that $a^x = b$ where $a$ and $b$ belong to some group of order $N$ and where $a$ is a generator of the group.

You should use the `crt` function from the module `crypto_basic`, which is available in the file `crypto_basic.py`.

### Plan:

1. Find group order
2. Find prime power factors
3. Set $g_i = g^{N/q_{i}^{e_i}}$, for $1 \leq i \leq r$
4. Set $h_i = h^{N/q_{i}^{e_i}}$, for $1 \leq i \leq r$
5. For each $1, ..., r$, set $G_i = \langle g_i \rangle$
6. Solve DLP $g_{i}^{x_i} = h_i$.
7. Finally, using the Chinese Remainder Theorem, solve: $x \equiv x_1 (\text{mod} \, q_{1}^{e_1}), \,  ... \,, \,  x \equiv x_1 (\text{mod} \, q_{r}^{e_r})$ 

````python
def pohlig_hellman(g, h, f):
    """Given that g and h are both non-zero elements of the same finite field, this function uses
    Pohlig-Hellman to return z such that g^z = h, if such a z exists. The argument f is a function such that
    f(a, b, N) returns x for which a^x = b, where a is a generator of a group of order N that contains b"""

    # 1 - Find group order
    deg = g.field.modulus.degree()
    N = 2**deg - 1

    #2 - Find prime power factors
    qs = prime_power_factors(N)  # each of these is already given to it's respective power

    # 3 and 4 - Set gi and hi
    gs, hs = [], [] 
    g_prime, h_prime = g, h
    for q in qs:
        g = g_prime**(N // q)
        gs.append(g)
        h = h_prime**(N // q)
        hs.append(h)

    # 5 - set Gi

    # 6 - solve DLP for gi ** xi = hi, f is the solving algorithm
    xs = []
    for gi, hi, q in zip(gs, hs, qs):
        xs.append(f(gi, hi, q))
    
    # 7 - CRT
    return crt(xs, qs)    
````
